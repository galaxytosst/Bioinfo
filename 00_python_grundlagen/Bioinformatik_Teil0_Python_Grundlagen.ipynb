## So benutzt du das Notebook
- Ein Notebook besteht aus **Zellen**. Text-Zellen (wie diese) erklären, **Code-Zellen** (grau) enthalten Python.
- Eine Code-Zelle führst du mit **Shift + Enter** aus. Die Ausgabe erscheint darunter.
- Führe die Zellen **von oben nach unten** aus. Spätere Zellen brauchen oft Dinge aus früheren.
- Wo `# TODO` steht, schreibst **du** den Code.
- Danach kommt eine **Prüf-Zelle** (🔍). Die führst du einfach aus: ✅ heißt richtig, ❌ sagt dir, was noch nicht stimmt.
- Unter jeder Übung gibt es eine aufklappbare **Lösung**. Erst selbst probieren, dann vergleichen. Es gibt fast immer mehrere richtige Lösungen.
- Fehlermeldungen sind normal und kein Drama. Die letzte Zeile der Meldung sagt meist, was los ist (z. B. `NameError: name 'dna' is not defined` = du hast eine Zelle weiter oben noch nicht ausgeführt).

**Führe als Erstes die nächste Zelle aus.** Sie definiert die Prüf-Werkzeuge. Du musst sie nicht verstehen.

````python
# Hilfswerkzeug zum Prüfen deiner Lösungen. Einfach ausführen.
import math

def pruefe(beschreibung, berechne, erwartet):
    try:
        ergebnis = berechne()
    except NameError as fehler:
        print(f"❌ {beschreibung}: Etwas ist noch nicht definiert ({fehler}). Hast du die Zelle darüber ausgeführt?")
        return
    except Exception as fehler:
        print(f"❌ {beschreibung}: Dein Code hat einen Fehler ausgelöst: {type(fehler).__name__}: {fehler}")
        return
    if isinstance(erwartet, float) and isinstance(ergebnis, (int, float)):
        richtig = math.isclose(ergebnis, erwartet, abs_tol=0.01)
    else:
        richtig = (ergebnis == erwartet)
    if richtig:
        print(f"✅ {beschreibung}: richtig!")
    elif ergebnis is None:
        print(f"❌ {beschreibung}: Ergebnis ist None. Noch nicht gelöst, oder fehlt ein 'return'?")
    else:
        print(f"❌ {beschreibung}: erwartet {erwartet!r}, bekommen {ergebnis!r}")

print("Prüf-Werkzeug geladen. Los geht's!")
````

## Python: Variablen und Strings
- Eine **Variable** ist ein Name für einen Wert: `dna = "ATGC"` speichert den Text `"ATGC"` unter dem Namen `dna`.
- Text in Anführungszeichen heißt **String**.

````python
dna = "ATGCGTAGGC"
anzahl_G = 0                     # Zähler startet bei 0

for base in dna:                 # 'base' ist nacheinander "A", "T", "G", ...
    if base == "G":
        anzahl_G = anzahl_G + 1  # Zähler um 1 erhöhen

print("G kommt", anzahl_G, "mal vor")
````

# Lektion 2: Transkription und GC-Gehalt

## Biologie: Transkription
Bei der Transkription liest die RNA-Polymerase den **Template-Strang** (Matrizenstrang) und baut eine
komplementäre mRNA von 5′ nach 3′. Ergebnis: Die mRNA hat **dieselbe Sequenz wie der codierende Strang**,
nur steht **U (Uracil) statt T**.

In der Bioinformatik gibt man Gene fast immer als codierenden Strang an. „Transkribieren“ heißt im Code
daher einfach: **jedes T durch U ersetzen**.

```
codierender Strang (DNA):  5′-ATGGCT-3′
mRNA:                      5′-AUGGCU-3′
```

## Python: Strings zusammenbauen
Strings kann man mit `+` aneinanderhängen: `"AT" + "G"` ergibt `"ATG"`.
Ein typisches Muster: mit einem leeren String `""` starten und in der Schleife Zeichen für Zeichen anhängen.
`else` heißt „in allen anderen Fällen“.

````python
# Beispiel: Alle C durch N ersetzen (N = "unbekannte Base", z. B. bei unsicheren Sequenzierdaten)
dna = "ACCGTCA"
neu = ""

for base in dna:
    if base == "C":
        neu = neu + "N"
    else:
        neu = neu + base

print(dna)
print(neu)
````

## Biologie: GC-Gehalt
Der **GC-Gehalt** ist der Anteil von G und C an allen Basen, meist in Prozent:

$$\text{GC-Gehalt} = \frac{\text{Anzahl G} + \text{Anzahl C}}{\text{Länge}} \cdot 100$$

Warum interessant? G–C-Paare bilden 3 Wasserstoffbrücken (A–T nur 2) und tragen stärker zur Basenstapelung bei.
GC-reiche DNA schmilzt daher bei höherer Temperatur (wichtig z. B. für PCR-Primer). Außerdem ist der GC-Gehalt
typisch für Organismen: Er reicht bei Bakterien grob von unter 20 % bis über 70 %.

## Python: Rechnen
`+ - * /` funktionieren wie erwartet. `/` liefert eine Kommazahl (Python schreibt einen **Punkt**: `0.5`).
`round(zahl, 2)` rundet auf 2 Nachkommastellen.

````python
a = 7
b = 3
print(a + b, a * b, a / b)
print(round(a / b, 2))
print(f"Ergebnis: {a / b:.2f}")   # f-String: Werte in Text einsetzen, :.2f = 2 Nachkommastellen
````

# Lektion 3: Reverse Complement

## Biologie
DNA ist ein Doppelstrang. Die Stränge sind **komplementär** (A↔T, G↔C) und **antiparallel**:
Der eine läuft 5′→3′, der andere 3′→5′.

```
5′-ATGCC-3′    gegebener Strang
3′-TACGG-5′    Gegenstrang
```

Sequenzen schreibt man immer 5′→3′. Den Gegenstrang 5′→3′ gelesen erhält man, indem man
**komplementiert und umdreht**: `GGCAT`. Das nennt man **Reverse Complement**.
Du brauchst es ständig: Gene können auf beiden Strängen liegen, Primer binden am Gegenstrang,
und Sequenzier-Reads kommen von beiden Strängen.

## Python: Dictionaries
Ein **Dictionary** (Wörterbuch) ordnet **Schlüsseln** Werte zu. Man schreibt es mit geschweiften Klammern
und schlägt mit eckigen Klammern nach:

````python
# Beispiel: Einbuchstaben-Code → Dreibuchstaben-Code von Aminosäuren
dreibuchstaben = {"M": "Met", "W": "Trp", "G": "Gly", "K": "Lys"}

print(dreibuchstaben["M"])
print(dreibuchstaben["K"])

protein = "MKGW"
for aminosaeure in protein:
    print(aminosaeure, "→", dreibuchstaben[aminosaeure])
````

**Umdrehen** eines Strings geht mit `text[::-1]` (eine sogenannte Slice-Schreibweise, mehr dazu in Lektion 4).

das diesen NCBI-Eintrag unverändert mitliefert.

## Das FASTA-Format
FASTA ist das einfachste und häufigste Format für Sequenzen:
```
>Kopfzeile mit Name und Beschreibung
ATGCGTACGTTAGC
GGCTAGCTAGGATC
>nächste Sequenz
...
```
- Eine Zeile, die mit `>` beginnt, ist die **Kopfzeile** (Header) eines neuen Eintrags.
- Alle Zeilen danach bis zum nächsten `>` gehören zur Sequenz. Sie ist auf mehrere Zeilen umgebrochen
  (hier: 70 Zeichen pro Zeile) und muss wieder zusammengesetzt werden.

## Dateien besorgen
Die nächste Zelle lädt zwei Dateien herunter (in Colab funktioniert das direkt):
- `pPCP1_gene.fasta`: die 10 Gene einzeln (je codierende Sequenz, auch die vom Gegenstrang schon in Leserichtung),
- `pPCP1_plasmid.fasta`: das komplette Plasmid.

Falls der Download nicht klappt: Lade die beiden Dateien aus dem Projekt über das **Ordnersymbol** links in Colab hoch.

````python
import os, urllib.request

basis = "https://raw.githubusercontent.com/biopython/biopython/master/Tests/GenBank/"
dateien = {"pPCP1_gene.fasta": "NC_005816.ffn", "pPCP1_plasmid.fasta": "NC_005816.fna"}

for lokal, online in dateien.items():
    if os.path.exists(lokal):
        print(f"{lokal} ist schon da.")
    else:
        try:
            urllib.request.urlretrieve(basis + online, lokal)
            print(f"{lokal} heruntergeladen.")
        except Exception as fehler:
            print(f"Download von {lokal} fehlgeschlagen ({fehler}). Bitte Datei manuell hochladen.")
````

## Python: Dateien lesen

def lese_fasta(dateiname):
    sequenzen = {}
    aktueller_name = None
    with open(dateiname) as f:
        for zeile in f:
            zeile = zeile.strip()
            if zeile == "":                 # leere Zeilen überspringen
                continue
            if zeile.startswith(">"):
                aktueller_name = zeile[1:]
                sequenzen[aktueller_name] = ""
            else:
                sequenzen[aktueller_name] = sequenzen[aktueller_name] + zeile
    return sequenzen

## 🔬 Bonus zum Nachdenken: nicht jedes Gen beginnt mit ATG
Die nächste Zelle zeigt das erste Codon jedes Gens und die erste Aminosäure nach unserer Übersetzung.

````python
for name, seq in gene.items():
    beschreibung = name.split(" ", 1)[1].split(" [")[0]
    print(seq[:3], "→", uebersetze(transkribiere(seq))[:1], " ", beschreibung)
````

Vier Gene beginnen mit **GTG** oder **TTG** statt ATG. Unsere Funktion übersetzt deren erstes Codon als V bzw. L.
NCBI gibt diese Proteine aber mit **M** am Anfang an. Wer hat recht?

Beide Codons sind in Bakterien **alternative Startcodons**. Steht ein Codon an der Startposition, wird es von der
Initiator-tRNA (fMet-tRNA) gelesen, und es wird immer **(Formyl-)Methionin** eingebaut. Mitten im Gen bedeutet GTG
dagegen ganz normal Valin. Unsere einfache Funktion kennt diese Sonderregel nicht.

*Wenn du Lust hast:* Erweitere `uebersetze` so, dass die erste Aminosäure immer `M` ist, wenn das erste Codon
`AUG`, `GUG` oder `UUG` ist.
